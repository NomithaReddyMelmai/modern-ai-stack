```python
# Bootstrap: make the repo root importable so `import config` works from notebooks/
import sys, os
sys.path.insert(0, os.path.abspath(".."))
from config import assert_key
assert_key()
print("Gateway ready.")
```

```python
from config import API_KEY, BASE_URL, FAST_MODEL, EMBED_MODEL
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
from llama_index.llms.openai_like import OpenAILike
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

Settings.llm = OpenAILike(model=FAST_MODEL, api_base=BASE_URL, api_key=API_KEY,
                          is_chat_model=True, is_function_calling_model=True, context_window=128000)
Settings.embed_model = HuggingFaceEmbedding(model_name=EMBED_MODEL)   # local, first run downloads ~90MB

docs = SimpleDirectoryReader("../data").load_data()
index = VectorStoreIndex.from_documents(docs)
qe = index.as_query_engine(similarity_top_k=3)
```

```python
resp = qe.query("How long are battery packs under warranty?")
print(resp)
print("sources:", [n.metadata.get("file_name") for n in resp.source_nodes])
```
