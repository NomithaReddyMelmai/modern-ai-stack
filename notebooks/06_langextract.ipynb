```python
# Bootstrap: make the repo root importable so `import config` works from notebooks/
import sys, os
sys.path.insert(0, os.path.abspath(".."))
from config import assert_key
assert_key()
print("Gateway ready.")
```

```python
import langextract as lx
from config import API_KEY, BASE_URL, FAST_MODEL

prompt = ("Extract each robot product mentioned, with its battery life and payload "
          "or key spec. Use exact text from the source; do not paraphrase.")
examples = [lx.data.ExampleData(
    text="Acme Falcon: aerial drone, 40-minute flight time, 2 kg payload.",
    extractions=[lx.data.Extraction(extraction_class="product", extraction_text="Acme Falcon",
                 attributes={"spec": "40-minute flight time", "payload": "2 kg"})])]

input_text = open("../data/company_faq.md").read()
result = lx.extract(text_or_documents=input_text, prompt_description=prompt, examples=examples,
                    model_id=FAST_MODEL, api_key=API_KEY, fence_output=True,
                    use_schema_constraints=False, language_model_params={"base_url": BASE_URL})
for e in result.extractions:
    print(f"[{e.extraction_class}] {e.extraction_text}  attrs={e.attributes}  span={e.char_interval}")
```
