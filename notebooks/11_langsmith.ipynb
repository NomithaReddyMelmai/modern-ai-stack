```python
import os
if os.environ.get("LANGSMITH_TRACING", "").lower() != "true":
    print("⚠️  LANGSMITH_TRACING is not 'true' in .env — set it to see traces.")

from config import get_langchain_llm
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langsmith import traceable

chain = ChatPromptTemplate.from_template("Summarize {topic} in one line.") | get_langchain_llm() | StrOutputParser()
print(chain.invoke({"topic": "vector databases"}))

@traceable(run_type="chain", name="two_step_pipeline")
def pipeline(topic: str) -> str:
    draft = chain.invoke({"topic": topic})
    return chain.invoke({"topic": f"rewrite more formally: {draft}"})

print(pipeline("observability for AI agents"))
print("Open LangSmith to see the traces.")
```
